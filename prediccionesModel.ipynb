```python
from xgboost import XGBClassifier
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
#Se carga el modelo previamente entrenado y guardado en un archivo JSON
model = XGBClassifier(enable_categorical=True)
model.load_model("modelo_xgb_v1.json")
```

```python
#Se carga el archivo CSV con los datos transformados
data_transformado = pd.read_csv("data_transformado.csv")
data_transformado.info()
```

```python
#Se prepara el conjunto de datos para la predicción
features = data_transformado.drop(columns=['target']).columns
X = data_transformado[features]
y = data_transformado['target']

columnas_object = X.select_dtypes(include=['object']).columns


for col in columnas_object:
    X[col] = X[col].astype('category')

seed = 48  # Semilla para reproducibilidad

X_train, X_val, y_train, y_val = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=seed, 
    stratify=y # Mantiene la proporción de clases (0 y 1) en train y test
)
```

```python
#Se verifica la información del conjunto de validación
X_val.info()
```

```python
#Se muestra los valores de un paciente seleccionado al azar
num_paciente = 0  # Índice del paciente a mostrar
print(X_val.iloc[num_paciente])
print("El paciente tiene enfermedad cardiaca: {}".format((y_val.iloc[num_paciente])))
```

```python
#Se realiza la predicción para un paciente específico del conjunto de validación
model.predict(X_val.iloc[[num_paciente]])
```
